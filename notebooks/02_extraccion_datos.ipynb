```python
import json
import re
import pandas as pd
import os

BASE = r'C:\Users\linoc\Desktop\CABATECH'

# Ruta correcta
ruta_html = r'C:\Users\linoc\Downloads\PropTech_CABA_V7.html'

with open(ruta_html, 'r', encoding='utf-8') as f:
    contenido = f.read()

# Extraer AZ (168 subzonas)
match = re.search(r'var AZ=(\[.*?\]);', contenido, re.DOTALL)
if match:
    az_json = match.group(1)
    az = json.loads(az_json)
    print(f"Subzonas extraídas: {len(az)}")
else:
    print("ERROR: no se encontró AZ")

# Extraer centroides
match2 = re.search(r'var CENT=(\{.*?\});', contenido, re.DOTALL)
if match2:
    cent = json.loads(match2.group(1))
    print(f"Centroides extraídos: {len(cent)}")

# DataFrame
df = pd.DataFrame(az)
print(f"Shape: {df.shape}")
print(df.columns.tolist())
```

```python
# Renombrar columnas a nombres legibles
columnas = {
    'b': 'barrio', 'z': 'subzona', 'pm': 'precio_m2', 're': 'rentabilidad',
    'roi': 'roi', 'vol': 'volatilidad', 'sh': 'sharpe', 'pp': 'prob_positivo',
    'v95': 'var95', 'sc': 'score', 'proy': 'proyeccion', 'pb': 'payback',
    'cz': 'categoria', 'em': 'emergente', 'curZ': 'zona_cur', 'curH': 'altura_max',
    'curP': 'pisos', 'cr': 'cap_rate', 'ff5': 'ff5', 'pis': 'n_pisos',
    'm2v': 'm2_vendibles', 'ctot': 'costo_total', 'ing': 'ingresos',
    'mrg': 'margen', 'prf': 'perfil', 'pvm': 'precio_venta_m2',
    'ipg': 'idx_gentrificacion', 'gcl': 'clase_gentrif', 'gv': 'gv',
    'gp': 'gp', 'gc': 'gc', 'gt': 'gt', 'sco': 'score_desarrollo',
    'dsub': 'dist_subte', 's5': 'subte_5min', 's10': 'subte_10min',
    'lsub': 'linea_subte', 'esub': 'estacion_subte', 'dhos': 'dist_hospital',
    'hcer': 'hospital_cercano', 'acc': 'accesibilidad'
}

df = df.rename(columns=columnas)

# Agregar coordenadas desde centroides
df['lat'] = df['barrio'].map(lambda b: cent.get(b, [None, None])[0])
df['lon'] = df['barrio'].map(lambda b: cent.get(b, [None, None])[1])

# Guardar CSV
ruta_csv = os.path.join(BASE, '168_subzonas_proptech.csv')
df.to_csv(ruta_csv, index=False, encoding='utf-8')
print(f"CSV guardado: {ruta_csv}")
print(df[['barrio', 'subzona', 'precio_m2', 'roi', 'score', 'idx_gentrificacion', 'categoria']].head(10))
```

```python
import geopandas as gpd
from shapely.geometry import Point

# Crear GeoDataFrame con puntos (centroides de barrio)
gdf = gpd.GeoDataFrame(
    df,
    geometry=[Point(lon, lat) for lon, lat in zip(df['lon'], df['lat'])],
    crs='EPSG:4326'
)

# Guardar GeoJSON
ruta_geojson = os.path.join(BASE, '168_subzonas_proptech.geojson')
gdf.to_file(ruta_geojson, driver='GeoJSON')
print(f"GeoJSON guardado: {ruta_geojson}")
print(f"Subzonas con coordenadas: {gdf['geometry'].notna().sum()}")
```

```python
import requests
import os

BASE = r'C:\Users\linoc\Desktop\CABATECH'
DATA = os.path.join(BASE, 'data')
os.makedirs(DATA, exist_ok=True)

# URLs directas de datos abiertos CABA
datasets = {
    'subte_estaciones.geojson': 'https://data.buenosaires.gob.ar/dataset/subte-estaciones/resource/bea83c40-9400-4fe4-865b-f0f0610a060c/download/subte-estaciones.geojson',
    'subte_lineas.geojson': 'https://data.buenosaires.gob.ar/dataset/subte-estaciones/resource/89fef3c6-c133-4eee-8612-510fad192e3a/download/subte-lineas.geojson',
}

for nombre, url in datasets.items():
    print(f"Descargando {nombre}...")
    r = requests.get(url, timeout=30)
    if r.status_code == 200:
        with open(os.path.join(DATA, nombre), 'wb') as f:
            f.write(r.content)
        print(f"✓ Guardado: {nombre} ({len(r.content)/1024:.1f} KB)")
    else:
        print(f"✗ Error {r.status_code}: {nombre}")
```

```python
# Verificamos qué bajamos realmente
for nombre in ['subte_estaciones.geojson', 'subte_lineas.geojson']:
    ruta = os.path.join(DATA, nombre)
    with open(ruta, 'r') as f:
        contenido = f.read()
    print(f"\n{nombre}:")
    print(contenido[:300])
```
